# 2 · Ranger une valeur

## 🔎 On regarde ensemble

Le `=` n'est pas un égal mathématique. C'est une **flèche** : « range
ce qui est à droite dans la boîte de gauche ».


````python
montant = 42.90
print(montant)

montant = 51.20        # l ancienne valeur est ECRASEE
print(montant)
````

Du coup `montant = montant + 10` n'a rien d'absurde : le côté droit est
calculé **d'abord**, puis rangé à gauche.


````python
montant = 42.90
montant = montant + 10
print(montant)

montant += 10          # la meme chose, en plus court
print(montant)
````

# 3 · Bien nommer

Les deux cellules suivantes calculent **exactement la même chose**.
Une seule se relit.


````python
# lisible par personne
m = 42.90
n = 3
r = m / n
print(r)
````

````python
# lisible par tous
addition = 42.90
convives = 3
part = addition / convives
print(part)
````

**La convention Python** : tout en minuscules, les mots séparés par des
tirets bas — `prix_unitaire`, `nombre_de_parts`. Ça s'appelle le
*snake_case*.

# 5 · Compter

## 🔎 On regarde ensemble

Sept opérateurs. Trois surprennent : `//`, `%` et `**`.


````python
print(7 + 2)      # 9
print(7 - 2)      # 5
print(7 * 2)      # 14
print(7 / 2)      # 3.5   <- toujours un decimal
print(7 // 2)     # 3     <- la division ENTIERE
print(7 % 2)      # 1     <- le RESTE de cette division
print(7 ** 2)     # 49    <- la puissance
````

> `7 / 2` rend `3.5`. Même `10 / 2` rend `5.0`, et non `5` :
> la division simple donne **toujours** un décimal.

# 7 · Assembler du texte

## 🔎 On regarde ensemble

Trois façons. Une seule est agréable.


````python
libelle = "Courses"
montant = 42.9

print(libelle + " : " + str(montant))    # ca marche, c est lourd
print(libelle, ":", montant)             # ca marche, peu de controle
print(f"{libelle} : {montant} EUR")      # la bonne facon
````

> Le `f` est **collé** au guillemet ouvrant. L'oublier n'est pas une
> erreur : ça affiche littéralement `{libelle}`. C'est l'oubli le plus
> fréquent de la séance.


## Mettre en forme un nombre

`:.2f` veut dire « deux chiffres après la virgule ». Sur de l'argent,
c'est la différence entre un programme et un brouillon.


````python
part = 42.90 / 3

print(f"Chacun paie {part} EUR")
print(f"Chacun paie {part:.2f} EUR")
````

# 8 · Le texte sait faire des choses

## 🔎 On regarde ensemble

On appelle une **méthode** avec un point. On reverra ce point en
séance 5 — pour l'instant, retiens la forme.


````python
saisie = "  courses du marche  "

print(saisie.strip())              # enleve les espaces autour
print(saisie.strip().upper())      # EN MAJUSCULES
print(saisie.strip().title())      # En Capitales
````

> `.strip()` sur **toute** saisie utilisateur : les gens tapent des
> espaces sans les voir, et `" courses "` n'est pas `"courses"` pour
> la machine.
