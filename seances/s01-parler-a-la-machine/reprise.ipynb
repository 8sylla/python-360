## 2. Les variables — la boîte étiquetée

`=` ne veut pas dire « égal ». Il veut dire « range ceci dans cette boîte ».


````python
prenom = "Awa"          # une boîte étiquetée "prenom" contenant du texte
age = 25                # un nombre entier
taille = 1.68           # un nombre à virgule
etudiant = True         # vrai ou faux

print(prenom, age, taille, etudiant)
print(type(prenom), type(age), type(taille), type(etudiant))
````

## 3. Le piège des types

On n'additionne pas des mots et des nombres.
La cellule suivante **provoque volontairement une erreur** : lis le message
jusqu'au bout, la dernière ligne dit toujours ce qui ne va pas.


````python
# Exécute, lis l'erreur, puis commente la ligne pour continuer.
resultat = "5" + 3
````

````python
# La bonne version : on convertit avant d'additionner.
resultat = int("5") + 3
print(resultat)
````
