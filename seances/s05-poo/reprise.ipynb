## 1. La classe = le moule, l'objet = le gâteau

```python
class Opportunite:
    """Une opportunité repérée."""

    def __init__(self, titre: str, pays: str, jours: int):
        # __init__ s'exécute à la création. self = "moi-même".
        self.titre = titre
        self.pays = pays
        self.jours = jours

    def est_urgente(self) -> bool:
        return self.jours < 7


bourse = Opportunite("Smarts-Up", "France", 12)
stage = Opportunite("Stage OCP", "Maroc", 3)

print(bourse.titre, "| urgente ?", bourse.est_urgente())
print(stage.titre, "| urgente ?", stage.est_urgente())
```

## 2. `@dataclass` : la même chose en trois lignes

```python
from dataclasses import dataclass, field


@dataclass
class OpportuniteSimple:
    titre: str
    pays: str
    jours: int
    tags: list[str] = field(default_factory=list)
    # ⚠️ default_factory=list et NON default=[] :
    # une liste par défaut serait PARTAGÉE par toutes les instances.


o = OpportuniteSimple("Smarts-Up", "France", 12)
print(o)        # __repr__ lisible, offert gratuitement
print(o == OpportuniteSimple("Smarts-Up", "France", 12))   # __eq__ aussi
```

## 3. `Enum` : la fin des chaînes magiques

```python
from enum import Enum


class Statut(str, Enum):
    A_FAIRE = "a_faire"
    EN_COURS = "en_cours"
    ENVOYEE = "envoyee"
    ARCHIVEE = "archivee"


print(Statut.A_FAIRE, "|", Statut.A_FAIRE.value)
print(Statut("envoyee"))
# Statut("envoyeee")   # décommente : une faute de frappe devient une ERREUR
```

## 4. `@property` : l'attribut qui se calcule tout seul

```python
from dataclasses import dataclass
from datetime import date, timedelta


@dataclass
class OpportuniteDatee:
    titre: str
    deadline: date

    @property
    def jours_restants(self) -> int:
        # Se lit SANS parenthèses et se recalcule à chaque lecture.
        return (self.deadline - date.today()).days


o = OpportuniteDatee("Test", date.today() + timedelta(days=5))
print(o.jours_restants, "jours")
```
