## 1. La fonction — la machine à café

```python
def saluer(prenom: str) -> str:
    """Rend un message de salutation.

    Cette ligne entre triples guillemets est une DOCSTRING :
    elle documente la fonction et s'affiche avec help(saluer).
    """
    return f"Bonjour {prenom} !"


message = saluer("Awa")
print(message)
help(saluer)
```

## 2. ⚠️ `print` contre `return` — la grande confusion

« `print` parle à l'humain. `return` parle au reste du programme. »


```python
def double_qui_affiche(n):
    print(n * 2)          # n'ARRIVE à rien : la fonction rend None

def double_qui_rend(n):
    return n * 2          # rend une valeur réutilisable

a = double_qui_affiche(5)
b = double_qui_rend(5)
print("a vaut", a, "| b vaut", b)
```

## 3. Les fichiers — la mémoire longue

```python
import json
from pathlib import Path

FICHIER = Path("demo.json")

donnees = [{"titre": "Test", "pays": "Maroc"}]

# "with" referme le fichier quoi qu'il arrive.
# encoding="utf-8" est OBLIGATOIRE pour les accents.
with FICHIER.open("w", encoding="utf-8") as f:
    json.dump(donnees, f, indent=2, ensure_ascii=False)

with FICHIER.open(encoding="utf-8") as f:
    print(json.load(f))
```

## 4. La ceinture de sécurité : `try / except`

```python
def demander_entier(question: str) -> int:
    """Redemande tant que la saisie n'est pas un entier."""
    while True:
        try:
            return int(input(question))
        except ValueError:      # on NOMME l'erreur attendue, jamais except nu
            print("  ⚠️  Merci de saisir un nombre entier.")


# demander_entier("Un nombre : ")   # décommente pour tester
```

## 5. `match / case` (Python 3.10+) — le menu

```python
choix = "2"

match choix:
    case "1":
        print("Ajouter")
    case "2":
        print("Afficher")
    case "3":
        print("Filtrer")
    case _:                    # le "sinon"
        print("Choix invalide")
```
