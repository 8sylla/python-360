## Section 9 — Mini-projet final : Gestionnaire de dépenses interactif v2
Cahier des charges :
1. Budget mensuel fixe de 300 EUR.
2. L'utilisateur saisit des dépenses une par une (`while`), jusqu'à taper `"fin"`.
3. Si une dépense dépasse le budget restant → alerte + refus de la compter (`continue`).
4. Si le budget restant tombe à 0 ou moins → arrêt complet de la saisie (`break`).
5. À la fin, classer le résultat : `"large"` (reste > 50), `"juste"` (0 < reste <= 50),
   `"depasse"` (reste <= 0).

````
budget = 300

while True:
    depense_str = input("Montant de la dépense (ou 'fin') : ")
    if depense_str == "fin":
        break

    montant = float(depense_str)

    # TODO : si le montant dépasse le budget restant, prévenir et NE PAS le compter
    if ...:
        print("Dépense refusée : dépasse le budget restant.")
        continue

    budget = budget - montant
    print("Budget restant :", budget)

    # TODO : si le budget atteint 0 ou moins, arrêter la boucle
    if ...:
        print("Budget épuisé, saisie arrêtée.")
        break

# TODO : classer le résultat final
if budget > 50:
    print("Situation : large")
elif ...:
    print("Situation : juste")
else:
    print("Situation : depasse")
````

## Section 10 — Auto-tests (sans `input`, logique pure)

````
def statut_budget(budget_initial, depenses):
    budget = budget_initial
    for m in depenses:
        if m > budget:
            continue
        budget = budget - m
        if budget <= 0:
            break
    if budget > 50:
        return "large"
    elif budget > 0:
        return "juste"
    else:
        return "depasse"

assert statut_budget(300, [50, 40, 30]) == "large", statut_budget(300, [50, 40, 30])
assert statut_budget(100, [60, 45]) == "depasse", statut_budget(100, [60, 45])
assert statut_budget(80, [50, 25]) == "juste", statut_budget(80, [50, 25])
print("Tous les tests passent ✅")
````
