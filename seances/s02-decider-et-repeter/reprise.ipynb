## 2. Le panneau de signalisation : `if / elif / else`

L'indentation (le décalage de 4 espaces) n'est pas décorative :
c'est elle qui dit « cette ligne est à l'intérieur du if ».


```python
jours_restants = 3

if jours_restants < 0:
    print("Deadline dépassée")
elif jours_restants < 7:
    print("URGENT")
else:
    print("Tu as le temps")
```

## 3. Répéter

`for` = je sais combien de fois. `while` = tant que la condition tient.

```python
# range(5) produit 0, 1, 2, 3, 4 : ça commence à 0 et s'arrête AVANT 5.
for i in range(5):
    print("Tour numéro", i)
```

```python
# Le motif "accumulateur" : une boîte qui grossit à chaque tour.
total = 0
for prix in [10, 25, 5]:
    total = total + prix     # équivaut à total += prix
print("Total :", total)
```

```python
# ⚠️ La boucle infinie : quelque chose DOIT changer à l'intérieur.
compteur = 0
while compteur < 3:
    print("compteur =", compteur)
    compteur = compteur + 1   # retire cette ligne et la boucle ne s'arrête jamais
```
