## On regarde ensemble : les faux amis

Un égal (`=`) sert à ranger. Deux égaux (`==`) servent à demander si c'est pareil.
De plus, une majuscule change tout, et un type aussi.

```
print("Loisirs" != "loisirs")
print(12 == "12")
print(12 == 12.0)
```

## On regarde ensemble : l'aiguillage avec IF / ELIF / ELSE

Le `if` est un aiguillage. Python s'arrête à la première vraie.

```
jours_restants = 2
if jours_restants < 0:
    print("Deadline dépassée")
elif jours_restants < 7:
    print("URGENT")
else:
    print("Tu as le temps")
```

## On regarde ensemble : for et range

On répète quand on sait combien de fois avec `for`. `range` s'arrête avant la borne de fin.

```
for i in range(3, 7):
    print(i)
```

## On regarde ensemble : while

On répète tant qu'une condition est vraie. Il faut changer la variable à l'intérieur, sinon boucle infinie.

```
n = 3
while n > 0:
    print(n)
    n = n - 1
```
