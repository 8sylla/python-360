## Atelier fameux N°1 — Le calendrier des dépenses (FizzBuzz)

Pour chaque jour du mois (1 à 30) :
- si le jour est un multiple de 3 et de 5 -> afficher "Jour chargé"
- sinon si multiple de 3 -> afficher "Jour de courses"
- sinon si multiple de 5 -> afficher "Jour de loyer"
- sinon -> afficher le jour

```python
# A toi de jouer
jour = int(input("Jour du mois : "))
if jour % 3 == 0 and jour % 5 == 0:
    print("Jour chargé")
elif jour % 3 == 0:
    print("Jour de courses")
elif jour % 5 == 0:
    print("Jour de loyer")
else:
    print(jour)
```

## Atelier fameux N°2 — La cagnotte d'Euler (Somme d'Euler #1)

Calcule la somme de toutes les dépenses (de 1 à 100 EUR, exclu 100) qui sont
des multiples de 3 ou de 5.

```python
# A toi de jouer
total = 0
for i in range(1,100):
  if i % 3 == 0 or i % 5 == 0:
    total += i
print(total)
```

## Exercice — Le nombre mystère

- L'ordinateur choisit un nombre entre 1 et 100
- Le joueur propose : « trop grand » ou « trop petit »
- Compter les essais et féliciter
- Bonus : limiter à 7 essais

```python
import random
secret = random.randint(1, 100)
user_input = int(input("devinez la valeur de l'ordinateur: "))
essais = 1;

while user_input != secret and essais <= 7:
  if user_input > secret:
    print("trop grand")
  else:
    print("trop petit")
  essais += 1
  user_input = int(input("devinez la valeur de l'ordinateur: "))

if essais <= 7:
  print(f"bravo, c'est bien {secret} au {essais}e essais!")
else:
  print(f"Vous avez raté, c'était {secret} !")

```

## Atelier 2 — Le contrôleur de dépenses

Une liste de six dépenses (en EUR) est donnée. Compte combien dépassent
50 EUR et affiche le total, puis la proportion.

```python
depenses = [12, 65, 8, 120, 30, 45]
a_risque = 0
total = 0

# A toi de jouer
for i in depenses:
  if i > 50:
    a_risque += 1
    total += i
print(f"Le nombre de depense à risque == {a_risque}")
print(f"Le total des depenses à risque == {total}")


```
