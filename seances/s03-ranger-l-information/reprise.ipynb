## 1. La liste — une liste de courses numérotée à partir de 0

```python
pays = ["Maroc", "France", "Sénégal", "Côte d'Ivoire"]

print(pays[0])      # le PREMIER (on compte à partir de 0)
print(pays[-1])     # le dernier, sans calculer la longueur
print(pays[1:3])    # de l'index 1 INCLUS à l'index 3 EXCLU
print(len(pays))    # combien d'éléments
print("Maroc" in pays)
```

```python
pays.append("Ghana")     # ajoute à la fin
pays.remove("France")    # retire par valeur
pays.sort()              # trie en place
print(pays)
```

## 2. Le dictionnaire — un répertoire : on cherche par NOM, pas par position

```python
opportunite = {
    "titre": "Bourse Smarts-Up",
    "organisme": "Université Paris Cité",
    "pays": "France",
    "jours": 12,
}

print(opportunite["titre"])
print(opportunite.get("ville"))            # None au lieu de planter
print(opportunite.get("ville", "inconnue"))

for cle, valeur in opportunite.items():
    print(f"{cle} : {valeur}")
```
