# 1 · Les Listes (Tableaux Dynamiques)

Une liste stocke des valeurs dans un ordre précis. En coulisses, c'est un **tableau dynamique de pointeurs** avec *sur-allocation*.
- **Complexité** : L'accès par index et l'ajout à la fin (`append`) sont ultra-rapides $\mathcal{O}(1)$.
- **Attention** : Insérer ou retirer au début (`insert(0, x)`) est très lent $\mathcal{O}(n)$, car Python doit décaler tous les éléments.
- **Slicing magique** : La syntaxe `[début:fin:pas]` permet d'extraire, mais aussi de modifier la liste en place.

```python
montants = [12.5, 45.0, 3.2, 120.0]
print("Premier montant :", montants[0])
print("Dernier montant :", montants[-1])

# Slicing : inversion sans copier
montants.reverse() # Modifie en place
print("Inversé :", montants)

# Ajouter un element
montants.append(8.5)
print("Apres ajout :", montants)
```

# 2 · Tuples et Ensembles (Sets)

### 🔒 Le Tuple (L'immutabilité performante)
Le Tuple est figé (immodifiable), sa taille est fixe.
- **Performance** : Python applique un recyclage mémoire très optimisé. Accès en $\mathcal{O}(1)$.
- **Astuce** : S'il contient une liste, cette liste interne *peut* être modifiée (immutabilité relative).
- **Attention** : Pour un tuple d'un seul élément, n'oublie pas la virgule : `(42,)`.

### 🎯 L'Ensemble / Set (Le roi de l'algèbre)
Un Set est un **dictionnaire sans valeurs** (table de hachage).
- **Performance** : La vérification d'appartenance (`x in set`) est **instantanée** en $\mathcal{O}(1)$, comparé à $\mathcal{O}(n)$ pour une liste.
- **Dédoublonnage** : `list(set(ma_liste))` est le moyen le plus rapide de supprimer les doublons d'une liste (mais on perd l'ordre).
- **Astuce** : La méthode `discard(x)` supprime un élément sans lever d'erreur s'il est absent.

```python
# Le Tuple : utiliser des parentheses
coordonnees = (48.85, 2.35)
lat, lon = coordonnees  # Unpacking (deballage)
print("Latitude :", lat)

# L'Ensemble (Set) : elimine les doublons a la vitesse de l'eclair
tags = ["resto", "transport", "resto", "loyer", "resto"]
tags_uniques = set(tags)
print("Tags uniques :", tags_uniques)

# Operations mathematiques sur les Sets
set_a = {1, 2, 3}
set_b = {3, 4, 5}
print("Union (|) :", set_a | set_b)
print("Intersection (&) :", set_a & set_b)
```

## Atelier 2 — Union des catégories

Tu as deux listes de catégories utilisées ce mois-ci et le mois dernier.
Transforme-les en `sets` et trouve :
1. Les catégories utilisées **dans les deux mois** (Intersection `&`).
2. **Toutes** les catégories utilisées au moins une fois sur les deux mois (Union `|`).

```python
mois_1 = ["Loyer", "Alimentation", "Loisirs", "Transport"]
mois_2 = ["Alimentation", "Loisirs", "Sante", "Shopping"]

# A toi de jouer :
set_mois_1 = set(mois_1)
set_mois_2 = set(mois_2)
print("Intersection :", set_mois_1 & set_mois_2)
print("Union :", set_mois_1 | set_mois_2)
```

# 3 · Les Dictionnaires (Puissance du Hachage)

Un dictionnaire associe une **clé** (un nom) à une **valeur** via une **table de hachage**.
- **Performance** : Accès, ajout, suppression en **temps constant $\mathcal{O}(1)$**.
- **Bonus 3.6+** : L'ordre d'insertion des clés est désormais préservé, avec une empreinte mémoire réduite.
- **Fusion (3.9+)** : L'opérateur `|` permet de fusionner deux dictionnaires facilement.

```python
depense = {
    "titre": "Pass Navigo",
    "montant": 84.10,
    "categorie": "Transport"
}

print("Titre :", depense["titre"])
depense["montant"] = 86.40  # Modification

# get() evite la fameuse KeyError si la cle n'existe pas
print("Nouveau montant :", depense.get("montant"))
print("Lieu :", depense.get("lieu", "Non precise"))

# Fusion de dictionnaires (Python 3.9+)
bonus = {"lieu": "Paris", "paye": True}
depense_complete = depense | bonus
print("Depense complete :", depense_complete)
```
