# 1 · La liste

## On regarde ensemble : les gestes de base

On compte à partir de zéro. La tranche s'arrête avant la borne de fin.

```
depenses = ["Loyer", "Courses", "Abonnement"]

print(depenses[0])       # Loyer — la premiere
print(depenses[-1])      # Abonnement — la derniere
print(depenses[1:3])     # ['Courses', 'Abonnement']
print(len(depenses))
print("Loyer" in depenses)
```

## On regarde ensemble : modifier une liste

Quatre gestes qui couvrent 90 % des besoins.

```
d = ["Loyer", "Courses"]
d.append("Abonnement")   # ajoute a la fin
d.remove("Loyer")        # retire par valeur
d.sort()                 # trie sur place, ne renvoie rien
print(d)
```

# 2 · Le tuple (n-uplet)

## On regarde ensemble

Un tuple se lit comme une liste, mais ne se modifie jamais après sa création.

```
ligne_ticket = ("Cafe", 3.50)

nom, prix = ligne_ticket   # dissociation : deux boites, un seul tuple
print(nom, prix)

# ligne_ticket[1] = 4.00  # decommente pour voir le TypeError
```

## Liste ou tuple ?

- Un panier qu'on remplit et qu'on vide -> **liste**
- Des coordonnées GPS, une date de naissance -> **tuple**

# 3 · Le dictionnaire

## On regarde ensemble

Une clé, une valeur. On cherche par nom, pas par position.

```
depense = {
    "libelle": "Facture EDF",
    "categorie": "Logement",
    "montant": 89.50,
}

print(depense["libelle"])
depense["montant"] = 92.10   # on met a jour
print(depense)
```

## La clé qui n'existe pas

`.get()` ne plante jamais — c'est le réflexe dès qu'une donnée vient de l'extérieur.

```
# print(depense["justificatif"])  # decommente pour voir le KeyError

print(depense.get("justificatif"))              # None
print(depense.get("justificatif", "manquant"))   # manquant
```

# 4 · L'ensemble (set)

## On regarde ensemble

Un ensemble ne garde jamais deux fois la même valeur.

```
categories = {"Loyer", "Courses", "Loyer", "Loisirs"}
print(categories)          # le doublon a disparu
print("Loyer" in categories)

a = {"Loyer", "Courses"}
b = {"Courses", "Loisirs"}
print(a & b)   # intersection
print(a | b)   # union
print(a - b)   # difference
```

## Le piège des accolades vides

`{}` ne crée pas un ensemble vide — ça crée un dictionnaire vide.

```
print(type({}))       # <class 'dict'>
print(type(set()))    # <class 'set'>
```

## On regarde ensemble

Une liste de dictionnaires, c'est déjà un tableau.

```
depenses = [
    {"libelle": "Loyer", "montant": 650, "payee": True},
    {"libelle": "Courses", "montant": 45, "payee": False},
    {"libelle": "Abonnement", "montant": 12, "payee": False},
]

for d in depenses:
    if not d["payee"] and d["montant"] > 20:
        print(f"A payer : {d['libelle']} ({d['montant']} EUR)")
```

## Bonus — collections.Counter

```
from collections import Counter

categories = ["Logement", "Alimentation", "Logement",
              "Loisirs", "Alimentation", "Alimentation"]

Counter(categories).most_common(2)
```
